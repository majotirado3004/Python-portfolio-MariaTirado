# Creating a flag that sorts products

```python
import pandas as pd
import os

path = '/Users/mariatirado/29-10-2025 Instacart Basket Analysis/02 Data/Prepared Data'

ords_prods_merge = pd.read_parquet(os.path.join(path, 'ords_prods_merge.parquet'))

print(ords_prods_merge.shape)
ords_prods_merge.head()
```

```python
df = ords_prods_merge[:1000000]
print(df.shape)
```

```python
# Defining function 

def price_label(row):

  if row['prices'] <= 5:
    return 'Low-range product'
  elif (row['prices'] > 5) and (row['prices'] <= 15):
    return 'Mid-range product'
  elif row['prices'] > 15:
    return 'High range product'
  else: return 'Not enough data'
```

```python
df['price_range'] = df.apply(price_label, axis=1)
```

```python
df['price_range'].value_counts(dropna = False)
```

# If-Statements with the loc() Function

```python
df = df.copy()
```

```python
df.loc[df['prices'] > 15, 'price_range_loc'] = 'High-range product'
```

```python
df.loc[(df['prices'] <= 15) & (df['prices'] > 5), 'price_range_loc'] = 'Mid-range product' 
```

```python
df.loc[df['prices'] <= 5, 'price_range_loc'] = 'Low-range product'
```

```python
df.loc[:, 'price_range_loc'].value_counts(dropna=False)
```

# If-Statement with loc() for ords_prods_merge

# If-Statements with For-Loops

```python
ords_prods_merge['order_dow'].value_counts(dropna = False)
```

```python
result = []

for value in ords_prods_merge['order_dow']:
  if value == 0:
    result.append('Busiest day')
  elif value == 4:
    result.append('Least busy')
  else:
    result.append('Regularly busy')
```

```python
result
```

```python
ords_prods_merge['day_type'] = result
```

```python
ords_prods_merge['day_type'].value_counts(dropna=False)
```

# “Busiest day” to  “Busiest days” (plural)

```python
ords_prods_merge['order_dow'].value_counts()
```

```python
import numpy as np

ords_prods_merge['busiest_days_new'] = np.where(
    ords_prods_merge['order_dow'].isin([0, 1]), 'Busiest days',  # top 2
    np.where(
        ords_prods_merge['order_dow'].isin([3, 4]), 'Slowest days',  # bottom 2
        'Regularly busy'))
```

```python
ords_prods_merge['busiest_days_new'].value_counts(dropna=False)
```

# Answer
According to the results:
Busiest days (0 and 1) → 11,874,777 orders, showing these are the two most active shopping days.

Slowest days (3 and 4) → 7,630,865 orders, confirming they have the least shopping activity.

Regularly busy days → 12,926,818 orders, representing the remaining days with steady traffic.

The difference between the busiest and slowest days suggests strong weekly purchase patterns — likely linked to weekend and start-of-week shopping habits.

These insights can help optimize promotions and staff scheduling, focusing resources on the high-demand days.

# Busiest hours of the day

```python
ords_prods_merge['order_hour_of_day'].value_counts(dropna=False)
```

```python
import numpy as np

ords_prods_merge['busiest_period_of_day'] = np.where(
    (ords_prods_merge['order_hour_of_day'] >= 17) & (ords_prods_merge['order_hour_of_day'] <= 23), 'Most orders',
    np.where(
        (ords_prods_merge['order_hour_of_day'] >= 7) & (ords_prods_merge['order_hour_of_day'] <= 16), 'Average orders',
        'Fewest orders'))
```

```python
ords_prods_merge['busiest_period_of_day'].value_counts(dropna=False)
```

```python
ords_prods_merge.groupby('busiest_period_of_day')['order_hour_of_day'].unique()
```

# Answer
The “Most orders” period occurs between 19:00 and 23:00, confirming that evenings are the busiest hours for Instacart users.

The “Average orders” period covers 07:00–16:00, showing steady activity throughout daytime hours.

The “Fewest orders” period happens between 00:00 and 06:00, when demand is significantly lower.

Most Instacart orders are placed during the Average orders window (≈23.7M orders), followed by the Most orders period (≈7.8M), and very few during Fewest orders (≈0.9M).

These insights can help optimize server capacity and staffing, reducing the risk of app slowdowns during the evening peaks.
