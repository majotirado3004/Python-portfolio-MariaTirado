```python
# Cleaned customers file (from your last step)
customers = pd.read_pickle(os.path.join(path, '02 Data', 'Prepared Data',
                                        'customers_cleaned.pkl'))
```

```python
customers.head()
```

# Making sure the join keys match in type

```python
ords['user_id'] = ords['user_id'].astype('int64', errors='ignore')
customers['user_id'] = customers['user_id'].astype('int64', errors='ignore')
```

# Ensuring one row per user in customers

```python
# If any duplicates slipped in, keep the first
customers = customers.drop_duplicates(subset='user_id')
```

# Merge on user_id

```python
ords_cust = ords.merge(customers, on='user_id', how='left')
```
